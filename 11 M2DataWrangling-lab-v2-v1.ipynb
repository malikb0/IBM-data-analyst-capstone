### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
# i have to summerize the dataset  
print(df.info())
print(df.describe())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# i have to generate basic statistics for numeric columns in the dataset
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns
print(numeric_columns)
print(df[numeric_columns].describe())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# i have to identify inconsistent data in the 'Country' and 'EdLevel' columns
print("Unique values in 'Country' column:")
print(df['Country'].unique())
print("\nUnique values in 'EdLevel' column:")
print(df['EdLevel'].unique())
```

```python
# i have to replace the missing values in the 'Country' column with 'Unknown'
df['Country'] = df['Country'].fillna('Unknown')
# i have to replace the missing and nan values in the 'EdLevel' column with 'Unknown'
df['EdLevel'] = df['EdLevel'].fillna('Unknown')
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# i have to standardize the 'Country' column  
df['Country'] = df['Country'].str.strip().str.title()
print(df['Country'].unique())
# i have to standardize the 'EdLevel' column
df['EdLevel'] = df['EdLevel'].str.strip().str.title()
print(df['EdLevel'].unique())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# i have to encode the 'Employment' column using one-hot encoding and store the result in a new DataFrame 'employment_dummies'
employment_dummies = pd.get_dummies(df['Employment'], prefix='Employment')
print(employment_dummies.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# i have to identify columns with the highest number of missing values  
missing_values_count = df.isnull().sum()
missing_values_percentage = (missing_values_count / len(df)) * 100 
missing_values_df = pd.DataFrame({'Column': df.columns, 'MissingValuesCount': missing_values_count, 'MissingValuesPercentage': missing_values_percentage})
missing_values_df = missing_values_df.sort_values(by='MissingValuesCount', ascending=False)
print(missing_values_df.head())
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# i have to impute missing values in column 'ConvertedCompYearly' with the mean of the column
mean_converted_comp_yearly = df['ConvertedCompYearly'].mean()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(mean_converted_comp_yearly)
print(df['ConvertedCompYearly'].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# i have to impute missing values in categorical column 'RemoteWork' with most frequent value
remote_work_counts = df['RemoteWork'].value_counts()
print("Value counts for 'RemoteWork' column:")
print(remote_work_counts)

mode_remote_work = df['RemoteWork'].mode()[0]
print(f"Mode of 'RemoteWork' column: {mode_remote_work}")

df['RemoteWork'] = df['RemoteWork'].fillna(mode_remote_work)

remote_work_counts = df['RemoteWork'].value_counts()
print("Value counts for 'RemoteWork' column:")
print(remote_work_counts)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here

min_converted_comp_yearly = df['ConvertedCompYearly'].min()
max_converted_comp_yearly = df['ConvertedCompYearly'].max()
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_converted_comp_yearly) / (max_converted_comp_yearly - min_converted_comp_yearly)
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_MinMax']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import numpy as np 
## Write your code here
# Apply log-transform (log(1+x))
df["ConvertedCompYearly_log"] = np.log1p(df["ConvertedCompYearly"])
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
# Replace special strings with numeric equivalents
df["YearsCodePro"] = df["YearsCodePro"].replace({
    "Less than 1 year": "0",
    "More than 50 years": "50"
})
# Convert to numeric, coercing errors to NaN
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
# Fill missing values with 0 (or another strategy if you prefer)
df["YearsCodePro"] = df["YearsCodePro"].fillna(0)
```

```python
bins = [0, 2, 5, 10, 20, 50, np.inf]
labels = ["Beginner", "Intermediate", "Experienced", "Senior", "Expert", "Veteran"]

df["ExperienceLevel"] = pd.cut(
    df["YearsCodePro"],
    bins=[0, 2, 5, 10, 20, 50, np.inf],
    labels=["Beginner", "Intermediate", "Experienced", "Senior", "Expert", "Veteran"],
    right=False,   # <-- intervals are [a, b)
    include_lowest=True
)
```

```python
print(df['ExperienceLevel'].value_counts())
```
