## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows
1. Count the number of duplicate rows in the dataset.
3. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
# 1. count the number of duplicate rows in the dataset
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
# 2. Display the first few duplicate rows
duplicate_rows = df[df.duplicated()]
print("First few duplicate rows:")
print(duplicate_rows.head())
```

### Task 2: Analysis of Duplicate Response Patterns
1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
2. Clarify that these represent multiple respondents with identical answers rather than duplicate records. Analyse which other columns frequently share identical values within these response-pattern groups.
   


```python
## Write your code here
# 1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
# 2. Clarify that thse represent multiple respondents with identical answers rather than duplicate records. 
# Analyse which other columns frequently share identical values within these response-pattern groups. 
duplicates = df[df.duplicated(subset=['MainBranch', 'Employment', 'RemoteWork'], keep=False)]
print("Duplicate response patterns based on MainBranch, Employment, and RemoteWork:")
print(duplicates.head())
patterns = duplicates.groupby(['MainBranch', 'Employment', 'RemoteWork']).size().reset_index(name='Count')
print("Response patterns and their counts:")
print(patterns)
duplicate_patterns = patterns[patterns['Count'] > 1]
print("Duplicate response patterns:")
print(duplicate_patterns)
duplicate_groups = duplicates.groupby(['MainBranch', 'Employment', 'RemoteWork'])
for name, group in duplicate_groups:
    print(f"Response pattern: {name}")
    print(group.head()) 
    
```

### Task 3: Visualize Shared Response Patterns
1. Create visualizations to show the distribution of shared response patterns across different categories.
2. Use bar charts or pie charts to represent the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment.


```python
## Write your code here
"""
1. Create visualizations to show the distribution of shared response patterns across different categories.
2. Use bar charts or pie charts to represent the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment.
 
"""
# create a bar chart to show the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment
pattern_counts = duplicates.groupby(['Country', 'Employment']).size().reset_index(name='Count')
# Arrange the data for better visualization
pattern_counts = pattern_counts.sort_values(by='Count', ascending=False)
# Filter data to show only patterns with more than one occurrence
pattern_counts = pattern_counts[pattern_counts['Count'] > 300]

plt.figure(figsize=(12, 6))
plt.bar(pattern_counts['Country'] + ' - ' + pattern_counts['Employment'], pattern_counts['Count'], color='skyblue')
plt.xlabel('Country and Employment')   
plt.ylabel('Count of Duplicate Response Patterns')
plt.title('Distribution of Duplicate Response Patterns by Country and Employment having more than 100 occurrences')
plt.xticks(rotation=90)
plt.show()


# create a pie chart to show the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment
pattern_counts = duplicates.groupby(['Country', 'Employment']).size().reset_index(name='Count')
# Filter data to show only patterns with more than one occurrence
pattern_counts = pattern_counts[pattern_counts['Count'] > 300]
plt.figure(figsize=(8, 8))
plt.pie(pattern_counts['Count'], labels=pattern_counts['Country'] + ' - ' + pattern_counts['Employment'], autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Duplicate Response Patterns by Country and Employment having more than 100 occurrences')
plt.axis('equal')   
plt.show()


```
